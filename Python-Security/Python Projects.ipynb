```python
# Project 1 (Module 4): Automated Password Strength Checker + Generator
# This matches one of the official projects in your PDF.
# Task:
# Build a single Python file that:

# Lets the user choose between generating a strong password OR checking an existing password.
# Uses your earlier password_check() logic.
# Uses your generate_strong_password() function.
# Saves the generated password to a file generated_passwords.txt (append mode).


import random
import string


def password_strength_check(pswd):
    
   
        
        errors = []
        
        if len(pswd)<8:
            errors.append("- at least 8 characters")
        if not any(char.isdigit() for char in pswd):
            errors.append("- at least one digit")
        if not any(char.islower() for char in pswd):
            errors.append("- at least one lowercase")
        if not any(char.isupper() for char in pswd):
            errors.append("- at least one upper case")
        if all(char.isalnum() for char in pswd):
            errors.append("- at least one special character")
    
        if errors:
            print("Password is not strong enough. Your password must include-\n")
            for error in errors:
                print(f"-{error}")
            return False
            
        if not errors:
            print(f"New password is set: {pswd}")
            with open("pswd_manual.txt", "w") as file:
                file.write(pswd)
            with open("pswd_manual.txt", "r") as file:
                pswd_txt = file.read()
                print(f"Your password is saved in pswd_manual.txt: {pswd_txt}")
            return True
            

def strong_password_generator(length=12):

    password = [random.choice(string.ascii_lowercase),
                random.choice(string.ascii_uppercase),
                random.choice(string.digits),
                random.choice(string.punctuation)]

    password += random.choices((string.ascii_letters + string.digits + string.punctuation), k=(length-4))

    random.shuffle(password)
    pswd = "".join(password)
    print(f"Your new password is: {pswd}")
    with open("pswd_auto.txt", "w") as file:
        file.write(pswd)
    with open("pswd_auto.txt", "r") as file:
        pswd_txt = file.read()
        print(f"Your password is saved in pswd_manual.txt: {pswd_txt}")



#=======Main command line============

choice = input("Do you want to set the password yourself? y/n").lower()

if choice == "y":
    while True:
        pswd = input("Enter your new password :\n")
        if password_strength_check(pswd):
            print("Your new password is: ", pswd)
            with open("pswd.manual.txt", "w") as file:
                file.write(pswd)
            with open("pswd.manual.txt", "r") as file:
                password = file.read()
            break
        else:
            input("Try again")
    
else:
    strong_password_generator(length=12)
    
```

```python
# Project Name: Simple Security Log Analyzer
# Goal:
# Given a security log, automatically extract:

# All valid IP addresses
# All email addresses
# All URLs

import re

log = """[2026-04-16 10:15] Login from 192.168.1.10 successful
[2026-04-16 10:16] Failed login from 999.999.999.999
[2026-04-16 10:17] Suspicious email: hacker@evil.net
[2026-04-16 10:18] Click here: https://malware123.xyz/download
[2026-04-16 10:19] Connection from 172.16.5.20
[2026-04-16 10:20] Support ticket from admin@company.com"""

valid_ips = re.findall(r"\b(?:25[0-5]|2[0-4][0-9]|[0-1]?[0-9][0-9]?)\.(?:25[0-5]|2[0-4][0-9]|[0-1]?[0-9][0-9]?)\.(?:25[0-5]|2[0-4][0-9]|[0-1]?[0-9][0-9]?)\.(?:25[0-5]|2[0-4][0-9]|[0-1]?[0-9][0-9]?)\b", log)

valid_emails = re.findall(r"[a-zA-Z0-9.-]+@[a-zA-Z]+\.[a-zA-Z]{2,}", log)

valid_URLs = re.findall(r"https?://[a-zA-Z0-9-.]+\.[a-zA-Z]{2,}(?:/[a-zA-Z0-9~!@#$%^&*()_+;./<>?:[\],]*)?", log)

print("Security log analysis report\n==================\n")

print(f"\nThe valid ips are:  {len(valid_ips)}")
for ip in valid_ips:
    print(f"   -{ip}")

print(f"\nThe valid URLs are: {len(valid_URLs)}")
for url in valid_URLs:
    print(f"   -{url}")

print(f"\nThe valid emails are: {len(valid_emails)}")
for email in valid_emails:
    print(f"   -{email}")
```
