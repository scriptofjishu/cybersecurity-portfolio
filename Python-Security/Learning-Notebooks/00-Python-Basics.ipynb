```python
#missing letters
#count function
#append function
#join function
#creating an empty list

attendance=[False]*26

sentence="the quick brown fox jumped over the cat"

for letter in sentence:
    if "a"<= letter<="z":
        position=ord(letter)-ord("a")
        attendance[position]=True
        print (letter, position)
print()
print(attendance)

all_present=all(attendance)
print("all present:", all_present)

lower_case=sum(attendance)
print("number of lower case letter is:", lower_case)

space_count=sentence.count(" ")               #count function
print("total number of spaces in sentence is:", space_count)

missing_letters=[]   #This line creates an empty list and gives it the name missing_letters.
                        #Why do we need this empty list?
                            #We need a place to collect/store all the letters that are missing from the sentence.
                                #Think of it like this:
                                    # Imagine you have a small basket (the list).
                                   # At the start, the basket is empty → []
                                    #As you check each letter a-z:
                                    #If a letter is present → you do nothing (leave the basket alone).
                                    #If a letter is missing → you put that letter into the basket (using .append()).
                                        #At the end:
                                        #The basket (missing_letters) will contain only the missing letters.
                                        #Then we can print them all nicely.
for i in range(26):
    if not attendance[i]:
        missing_letter=chr(i+ord("a"))
        missing_letters.append(missing_letter)
if missing_letters:    ## [] is "falsy" → means empty
    print("missing_letters:", ",".join(missing_letters))
else:
    print("its a perfect anagram")
```

```python
# Task: Calculate how many days until the Great Migration

# 1. Import datetime
# 1. 导入 datetime 模块
import datetime

# 2. Get today's date
# Use: datetime.date.today()
today = datetime.date.today() 
print("Today is " + str(today))

# 3. Create the destination date (The Great Migration): 2030-01-01
# Use: datetime.date(2030, 1, 1)
migration_day = datetime.date(2026, 1, 1)
print("Migration date is " + str(migration_day))

# 4. Calculate the time remaining
# Subtract 'today' from 'migration_day' to get a timedelta
time_left = migration_day - today

# 5. Print the days remaining
# Use: time_left.days
# print("Days until migration: " + str(time_left.days))
print("Days until migration: " + str(time_left.days))
```
