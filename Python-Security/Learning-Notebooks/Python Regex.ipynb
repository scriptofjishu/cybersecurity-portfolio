```python
# \d matches any single digit (0 to 9).
# Small Example Sentence:
# "My phone number is 98765-43210 and my age is 25."

import re

sen = "My phone number is 98765-43210 and my age is 25."

dig = re.findall(r"\d", sen)

print(dig)
```

```python
# \d+ matches continuous whole numbers. matches one or more digits continuously (a chunk/group of digits)
# Small Example Sentence:
# "My phone number is 98765-43210 and my age is 25."

import re

sen = "My phone number is 98765-43210 and my age is 25."

dig = re.findall(r"\d+", sen)

print(dig)
```

```python
# \D matches any character that is NOT a digit.
# Small Example Sentence:
# "My phone number is 98765-43210 and my age is 25."

text = "My phone number is 98765-43210 and my age is 25."

not_digit = re.findall(r"\D", text)
not_digits = re.findall(r"\D+", text)

print(not_digit)

print("\n",not_digits)
```

```python
# \w matches any word character:

# Letters (a-z, A-Z)
# Digits (0-9)
# Underscore (_)

# In short: It matches alphanumeric characters + underscore.


text = "My phone number is 98765_43210 and my age is 25."

chars = re.findall(r"\w", text)
words = re.findall(r"\w+", text)

print(chars)
print(words)
```

```python
# \W matches any character that is NOT a word character.
# Opposite of \w.
# It matches:

# Spaces
# Punctuation (-, @, ., =, etc.)
# Special symbols


text = "My phone number is 98765-43210 and my age is @25!."

not_word = re.findall(r"\W", text)
not_words = re.findall(r"\W+", text)

print(not_word)
print(not_words)
```

```python
# \s (Whitespace) matches any whitespace character:

# Space ( )
# Tab (\t)
# Newline (\n)

text = """My phone number is 98765-43210 and my age is 25. 
Meaning:
\\s matches any whitespace character:

Space ( )
Tab (\\t)
Newline (\\n)"""

wspace = re.findall(r"\s", text)
wspaces = re.findall(r"\s+", text)

print(wspace)
print(wspaces)
```

```python
# . (Dot)
# The dot . matches any single character except a newline (\n).
# It is often called the wildcard character.

text = """First line of text.
Second line here.
Third line with number 123."""

result = re.findall(r".", text)
results = re.findall(r".+", text)   # → Will match each full line as one item (because .+ stops at newline).

print(result[:30])      # first 30 characters only
print(results)
print(results[:2])   # first 2 tuple only
```

```python
# . (Dot)


text = """First line of text.
Second line here.
Third line with number 123."""

result = re.findall(r".[:30]", text)
results = re.findall(r".+", text)

print(result[:30])      # first 30 characters only
print(results)
print(results[:2])
```

```python
# ^ (Caret) - Start of Line/String
# Meaning:
# ^ matches the beginning of a string or a line.


import re

text = """First line of text.
Second line here.
Third line with number 123.
Error: Something went wrong."""

result = re.findall(r"^\w+", text, re.MULTILINE)         #By default, ^ only matches the start of the entire string.
                                                          # When we add the flag re.MULTILINE, ^ starts matching at the beginning of every line.
                                                       

print(result)
```

```python
# $ (Dollar) - End of Line/String
# Meaning:
# $ matches the end of a string or a line.

import re 


text = """First line of text.
Second line here.
Third line with number 123.
Error: Something went wrong."""

result = re.findall(r"\w+$", text, re.MULTILINE)

print(result)


# Why did it return empty?
# Your pattern r"\w+$" means:

# \w+ → one or more word characters (letters, digits, underscore)
# $ → at the end of the line

# But look at your text:

# Most lines end with a period (.) → "text.", "here.", "123.", "wrong."

# The period . is not a word character, so \w+$ fails to match anything at the end of the lines.
```

```python
# [] - Character Set
# Meaning:
# [] lets you define a custom set of characters you want to match.
# Any one character inside the brackets will match.


text = "My phone: 98765-43210, email: jishu@python.com, age=25, score: 95%"

import re

result1 = re.findall(r"[a-z]", text)
print(result1)

result2 = re.findall(r"\d+", text)
print(result2)

result3 = re.findall(r"[aeiou]", text)
print(result3)

result4 = re.findall(r"[a-zA-Z]+", text)
print(result4)
```

```python
# [] Character Set (We already did some, but let's do a proper one)
# Meaning:
# [] lets you define your own set of allowed characters.


import re

text = "My score is 95%, grade: A+, rank #1, temp=37.5°C"

print(re.findall(r"[A-Z]+", text))           # Only uppercase letters
print(re.findall(r"[0-9]+", text))           # Only digits
print(re.findall(r"[0-9%+]+", text))         # Digits + % and +
print(re.findall(r"[A-Za-z0-9]+", text))     # Alphanumeric only
print(re.findall(r"[^a-zA-Z0-9]", text))     
print(re.findall(r"\W", text))
```

```python
# [^ ] → Negation (Not these characters)
# Meaning:
# [^ ] matches any character EXCEPT the ones you put inside the brackets.


text = "My phone: 98765-43210, email: jishu@python.com, age=25, score: 95%"

import re

result = re.findall(r"[^0-9]+", text)
print(result)

result2 = re.findall(r"[\W]+", text)  #collects only special characters
print(result2)
```

```python
# | (Pipe) - OR (Alternative)
# Meaning:
# | means OR. It allows you to match one pattern OR another.

import re

text = "I love cats and dogs and birds. I hate snakes."

result = re.findall(r"cat|dog|snake", text)
print(result)

log = "LOGIN SUCCESS, LOGIN FAILED, ACCESS DENIED"
result1 = re.findall(r"SUCCESS|FAILED|DENIED", log)
print(result1)
```

```python
# * (Asterisk)
# Meaning:
# * means zero or more occurrences of the previous character or group.
# (It is similar to +, but * allows zero matches too.)

text = "I love caaaats, dogsss, and birdddds. I saw a cat."

import re

result1 = re.findall(r"ca*t", text)
print(result1)

result2 = re.findall(r"dogs*", text)
print(result2)

# Match any word ending with 's' (zero or more letters before s)
result3 = re.findall(r"\w+s", text)
print(result3)

result4 = re.findall(r"\w+s*", text)
print(result4)
```

```python
# ? (Question Mark)
# Meaning:
# ? means zero or one occurrence of the previous character.

text = "color colour favorite favourite"

import re

result1 = re.findall(r"colou?r", text)
print(result1)

result2 = re.findall(r"favo(u?)rite", text)
print(result2)
```

```python
# {n} and {n,m} - Quantifiers (Exact Count)
# Meaning:

# {n} → Match the previous character/group exactly n times
# {n,m} → Match the previous character/group between n and m times (inclusive)


text = "My phone numbers are 98765, 1234567890, 555, 44, 987654321"

import re

print(re.findall(r"\d{5}", text))

print(re.findall(r"\d{3,4}", text))

print(re.findall(r"\d{10}", text))
```

```python
# ^ and $ (Start and End Anchors)
# Meaning:

# ^ → Matches at the start of a string or line
# $ → Matches at the end of a string or line


import re

text = """Error: System failed
Warning: Low memory
Success: Process completed
Error: Disk full"""

print(re.findall(r"^Error", text, re.M))

print(re.findall(r"(full|completed)$", text, re.M))

print(re.findall(r"^Warning.*memory$", text, re.M))
```

```python
# () - Capturing Groups (Very Important)
# Meaning:
# Round brackets () are used to capture (save) part of the match so you can extract it separately.


text = """My name is Jishu Sharma. My email is jishu2025@gmail.com. 
My phone is 98765-43210."""

import re

result1 = re.findall(r"(\w+)@[\w.]+", text)
print(result1)
# r"(\w+)@[\w.]+"
# Breakdown:

# ( → Start of capturing group (we want to save this part)
# \w+ → One or more word characters (letters, digits, underscore)
# ) → End of capturing group
# @ → Literal @ symbol
# [\w.]+ → One or more of (word character or dot)

# Why output = ['jishu2025']
# Only the part inside () was captured → jishu2025
# The rest (@gmail.com) was used for matching but not saved.

print(re.findall(r"\w+(@[\w.]+)", text))
print(re.findall(r"@[\w].+", text))
# r"@[\w.]+"
# Breakdown:

# @ → Literal @
# [\w.]+ → One or more of (word character or dot)

# Why output = ['@gmail.com.']
# It matched from @ till the end of the domain, including the final dot.

result2 = re.findall(r"(\w+@[\w.]+)", text)
print(result2)
# r"(\w+@[\w.]+)"
# Breakdown:

# ( → Start capturing group
# \w+ → One or more word characters (username)
# @ → Literal @
# [\w.]+ → One or more (word char or dot) → domain
# ) → End capturing group

# Why output = ['jishu2025@gmail.com.']
# The entire email is inside one capturing group, so the full email is returned.

print(re.findall(r"\w+@[\w.]+", text))
# when there is no capturing group at all, re.findall() returns the full match.

result3 = re.findall(r"name is (\w+ \w+).*phone is ([\d-]+)", text)
print(result3)

```

```python
How does .* work? ("skips everything in between")
This is a very important concept.

    
.* means:

. → Any character (except newline)
* → Zero or more times

So .* = "Match anything, as much as possible" (greedy)

Visual Example:
Pattern: name is (\w+ \w+).*phone is ([\d-]+)
Text:
name is Jishu Sharma. My email is jishu2025@gmail.com. My phone is 98765-43210.
How regex engine reads it:

Finds name is 
Captures Jishu Sharma
.* → eats everything until it can match the next part phone is 
Captures 98765-43210

The .* acts like a bridge — it skips over all the junk in the middle (email, url, etc.) to reach the next desired part.

Simple Analogy:
Think of .* as a greedy vacuum cleaner:

It sucks up everything it can until it finds the next required pattern.
```

```python
# Capturing Groups ()
# We already touched on this earlier, but let’s make it crystal clear.
# When you put normal parentheses () around part of a pattern, it creates a capturing group.
# What it does:

# It matches the pattern as usual.
# But it also "remembers" (captures) the part inside the parentheses.
# re.findall() will return only the captured part(s).

import re

log = "user=admin ip=192.168.1.10"

# Without capturing group
print(re.findall(r"user=[a-zA-Z0-9]+", log))
# Output: ['user=admin']

# With capturing group
print(re.findall(r"user=([a-zA-Z0-9]+)", log))
# Output: ['admin']
```

```python
# # Capturing vs Non-Capturing Groups (Real-life Log)
# Extract only the usernames (admin, guest123) using a capturing group.
# Extract full login lines (the entire line) using a non-capturing group for the timestamp part.
 
import re

log = """
[2026-04-16 10:15:22] LOGIN SUCCESS user=admin ip=192.168.1.10
[2026-04-16 10:16:05] LOGIN FAILED user=guest123 ip=10.0.0.1
[2026-04-16 10:17:11] SUSPICIOUS ACTIVITY ip=172.16.5.20 url=https://evil.com
"""

print(f"The usernames are {re.findall(r"user=(\w+)", log, re.M)}")

print(re.findall(r"^(?:\[[^\]]+\]\s*LOGIN FAILED)\s*(.*)", log, re.M))


```

```python
# Capturing vs Non-Capturing Groups (Real-life Log)
# Rule to Remember:
# When a capturing group ( ... ) is repeated with {n}, re.findall() only keeps the last match of that group.
# This is why we use non-capturing groups (?: ... ) when we repeat parts but want the full match.

# Extract IPs from Failed Login Lines
# Now we will improve the code to extract only the IP addresses from the failed login attempts (not from the whole log).
# Task:
# Modify your code to also extract the IP addresses only from the failed login lines.
# Hint: Use a capturing group around the IP part in the pattern.

import re

with open("security.log", "w") as f:
    f.write("""[2026-04-16 10:15:22] LOGIN SUCCESS user=admin ip=192.168.1.10
[2026-04-16 10:16:05] LOGIN FAILED user=guest ip=10.0.0.1
[2026-04-16 10:17:11] SUSPICIOUS ACTIVITY ip=172.16.5.20
[2026-04-16 10:18:33] EMAIL SENT to=hacker@evil.net
[2026-04-16 10:19:44] LOGIN FAILED user=admin ip=999.999.999.999""")

    
with open("security.log", "r") as f:
    security_log = f.read()
    suspicious_ip = re.findall(r".*LOGIN FAILED.*?(\b(\d{1,3}\.){3}\b)", security_log, re.I)  
    for ip in suspicious_ip:
        print(ip)


# Visual Breakdown with IP 999.999.999.999
# The regex engine does this:

# First repetition: captures 999.
# Second repetition: captures 999.
# Third repetition: captures 999.

# → Because the inner group (\d{1,3}\.) is repeated, only the last captured value (999.) is kept.

# You get a tuple:

# First element: What the outer group captured (999.999.999.)
# Second element: What the inner repeated group captured last (999.)
```

```python
# Capturing vs Non-Capturing Groups (Real-life Log)
# When a capturing group ( ... ) is repeated using {n}, re.findall() only remembers the last match of that group.

text = "Server IP: 192.168.1.10"

print(re.findall(r"(\d{1,3}\.){3}\d{1,3}", text))    # # Only last group captured
print(re.findall(r"(?:\d{1,3}\.){3}\d{1,3}", text))  # whole pattern is captured. () used before {3} to repeat the whole pattern. then without ?: 
                                                     # it would return only last repeatation of {3}, with ?: the group is non capturing, then  
                                                     #  returns the whole pattern.
```

```python
# Capturing Groups Repeated Words
import re

text = "hello hello hello world"

print(re.findall(r"(hello ){2}", text))   # Only last "hello " captured

print(re.findall(r"(?:hello ){2}", text))  # Full match
```

```python
r"^(?:\[[^\]]+\])\s*(.*)"
Part,Meaning,What it does in this pattern

^      Start of the line (because we use re.MULTILINE),Matches from the beginning of each line

(?:    Start of a non-capturing group,Groups the timestamp without saving it

\[     Literal [ character (escaped because [ has special meaning),Matches the opening bracket of timestamp

[^\]]+ One or more characters that are NOT ],Matches everything inside the brackets

\]     Literal ] character,Matches the closing bracket

)      End of non-capturing group,Ends the timestamp group

\s*    "Zero or more whitespace characters (space, tab, etc.)",Skips the space after the timestamp

(.*)   "Capturing group: Any character (except newline), zero or more times",Captures everything after the timestamp till end of line

(implied end),End of pattern,Stops at end of line
```

```python
Regex Flags
Now we will learn Regex Flags — the last major basic concept in regex.
What are Flags?
Flags are options that change the behaviour of the entire regex pattern.
They are passed as the third argument to functions like re.search(), re.findall(), etc.
Syntax:
re.findall(pattern, string, flag)

re.IGNORECASE or re.I
Makes the pattern case-insensitive.
A and a are treated as the same.

import re

log = """ERROR: Login failed
error: suspicious activity
Warning: weak password"""

print(re.findall(r"error", log))

print(re.findall(r"error", log, re.I))
```

```python
# re.IGNORECASE or re.I
# Makes the pattern case-insensitive.
# A and a are treated as the same.

import re

log = """ERROR: Login failed
error: suspicious activity
Warning: weak password"""

print(re.findall(r"error", log))

print(re.findall(r"error", log, re.I))
```

```python
# IGNORECASE Practical example

import re

log = """[2026-04-16 10:15] LOGIN SUCCESS user=admin
[2026-04-16 10:16] ERROR: Login failed for user=guest
[2026-04-16 10:17] Error: Suspicious activity detected
[2026-04-16 10:18] WARNING: Weak password used
[2026-04-16 10:19] error: Brute force attempt from IP 10.0.0.1"""

suspicious = re.findall(r"error|failed|warning", log, re.I)

print("Suspicious log found")

for event in suspicious:
    print("-", event)
```

```python
# # \b - Word Boundary
# # \b is a word boundary. It matches the position between a word character (\w) and a non-word character (\W), or at the start/end of a string.
# # It helps you match whole words only, not parts of words.
# \b ensures that the pattern is surrounded by non-word characters (or start/end of string), so it matches whole words only.
# You're right. \b (word boundary) works well for single words, but it becomes problematic when your pattern has spaces or is a phrase/sentence.
# Why \b fails with spaces:
# \b only works between a word character (\w) and a non-word character.
# When there are spaces, it often breaks the match.
    
import re

text = "I love cats and catnip. The category is cat. My cat is cute."

print(re.findall(r"\bcat\b", text))
print(re.findall(r"cat", text))
print(re.findall(r"\b\w*cat\w*\b", text))
```
