```python
#magic 8 ball


import random

def luckynumber(value):
    if value==1:
        return("good start")
    if value==2:
        return("cool")
    if value ==3:
       return("this is it")
    if value==4:
       return("maybe")
    if value==5:
        return("thats okay")
    if value==6:
        return("not it")
    if value==7:
        return("still not it")
    if value==8:
        return("wrong")

r=random.randint(1,9)

print(luckynumber(r))
```

```python
#diamond with star

n=int(input("enter diamond base"))

def diamond(n):
    stars=int(input("enter diamond base"))
    n=(stars//2)+2
    for i in range(1,n-1):
        print(" "*(n-i) + "".join(str("*") for j in range(1,1+i)) + "".join(str("*") for j in range(1,i)))
    for i in range(n-1, 0, -1):
        print(" "*(n-i) + "".join(str("*") for j in range(1,1+i)) + "".join(str("*") for j in range(1,i)))

finish=diamond(n)
print(finish)
```

```python
#diamond with stars


def diamond(stars):
    row=[]
    n=(stars//2)+1
    for i in range(1, n+1):
        upper=" "*(n-i) + "*"*((2*i)-1)
        row.append(upper)
    for i in range(n-1,0,-1):
        lower=" "*(n-i) + "*"*((2*i)-1)
        row.append(lower)
    return "\n".join(row)
print(diamond(9))
```

```python
#password checker


def check_password(password):
        errors=["password must contain:-"]
        if len(password)<8:
            errors.append("-at least 8 characters")
        if not any(char.isdigit() for char in password):
            errors.append("-at least one digit")
        if not any(char.islower() for char in password):
            errors.append("-at least one lower character")
        if not any(char.isupper() for char in password):
            errors.append("-at least one upper case")
        if all(char.isalnum() for char in password):
            errors.append("-at least one special character")
        if len(errors)==1:
            return ("password is strong")
        else:
            return "\n".join(errors)     
password=input("enter your password here\n")
print("\n" + check_password(password))
```

```python
#pasword genetor 1


import random
import string

l=random.randint(1,9)
lower=random.choices(string.ascii_lowercase, k=l)

u=10-l
upper=random.choices(string.ascii_uppercase, k=u)

s=11-l-u
special=random.choices(string.punctuation, k=s)

d=12-l-u-s
digits=random.choices(string.digits, k=d)

lists=lower + upper + special + digits
final=random.sample(lists, len(lists))
pswd="".join(final)

print(pswd)
```

```python
#password generator 2

import random
import string

l = random.choices(string.ascii_lowercase, k=1)
u = random.choices(string.ascii_uppercase, k=1)
d = random.choices(string.digits, k=1)
s = random.choices(string.punctuation, k=1)

pswd_list = l + u + d + s

catagories = [string.ascii_lowercase, string.ascii_uppercase, string.digits, string.punctuation]

for i in range(8):
    select_catagory=random.choice(catagories)
    char = random.choice(select_catagory)
    pswd_list.append(char)

print(pswd_list)

random.shuffle(pswd_list)
pswd = "".join(pswd_list)

print("your new password is: ", pswd)
```
